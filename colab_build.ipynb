# Compilador de NanoPal APK
Este notebook compilará el código fuente de NanoPal y generará el archivo APK listo para instalar.

### Instrucciones:
1. Sube el archivo `NanoPal.zip` a la sección de archivos de Colab (icono de carpeta a la izquierda).
2. Ejecuta todas las celdas (`Entorno -> Ejecutar todas`).
3. Al finalizar, el APK se descargará automáticamente.

```python
# 1. Instalar dependencias y SDK de Android
!apt-get update
!apt-get install -y openjdk-17-jdk-headless
import os
os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-17-openjdk-amd64"
```

```python
# 2. Descomprimir el proyecto
!unzip -o NanoPal.zip -d .
%cd NanoPal
!chmod +x gradlew
```

```python
# 3. Compilar APK
!./gradlew assembleDebug
```

```python
# 4. Descargar el resultado
from google.colab import files
apk_path = 'app/build/outputs/apk/debug/app-debug.apk'
if os.path.exists(apk_path):
    files.download(apk_path)
else:
    print("Error: No se encontró el APK generado.")
```
